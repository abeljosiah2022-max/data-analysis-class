```python
#Import libraries
import numpy as np
import pandas as pd
```

```python
#working with numpy array
arr = np.array([10, 20, 30, 40, 50, 60, 70])
print("1D Array:", arr)
print("Average Number:", arr.mean())
```

```python
#Indexing - for accessing an element on a list
print(arr[3])
print(arr[-2])
print(arr[4])
print(arr[6])
```

```python
#Slicing - for extracting a portion of a list
print(arr[1:6])
print(arr[6:])
print(arr[:7])
print(arr[-6:-1])
print(arr[7:])
print(arr[-1:-6])
```

```python
#30th April

#Creating array using np.arange 
arr3 = np.arange(1, 13)
print('Array3 Number:',arr3)

#Converting 1D array to 2D using 'Reshape'
reshape = arr3.reshape(4,3)
print('Reshaped 1D to 2D:\n', reshape)

#Converting 2D array to 1D using 'Flatten'
flatten = reshape.flatten()
print('Flatten Array:', flatten)
```

```python
#Maths Operators
arr4 = np.array([1, 2, 3, 4, 5])
arr5 = np.array([10, 20, 30, 40, 50])

#Basic Maths Operations
print('Addition:', arr4 + arr5)
print('Subtraction:', arr4 - arr5)
print('Multiplition:', arr4 * arr5)
print('Division:', arr4 / arr5)

#Scaler Operation
print('Exponential:', arr5**2)
print('Floor Div:', arr5//2)
print('Modulus:', arr5%3)
```

```python
#Dot Product using np.dot
matrix1 = np.array([
    [1, 2],
    [3, 4]
])
matrix2 = np.array([
    [5, 6],
    [7, 8]
])
dot_product = np.dot(matrix1, matrix2)
multiply = matrix1 * matrix2

print('Dot Product Result:\n', dot_product)
print('Product:\n', multiply)
```

```python
#30-day fitness tracker analysis
np.random.seed(42) #reproducibility it help generate the same set of random numbers
days = np.arange(1, 31)
steps = np.random.randint(3456, 12034, size=30)

print('Days:', days)
print('Steps:', steps)
```

```python
#Use pandas to convert the output to rows and columns using "pd.dataframe
data = pd.DataFrame({
    "Days": days,
    "Steps": steps
})

data
```

```python
#Basic Stats
print('Average Step', np.mean(steps))
print('Median Step', np.median(steps))
print('Minimum Step', np.min(steps))
print('Maximum Step', np.max(steps))
print('Step Variability', np.std(steps))
```

```python
#Goal analysis using np.where()
#Over 10000 Stepa
goal = np.where(steps > 10000)[0] + 1
print('Days with 10000+ Steps:', goal)

```

```python
goal_btw_1 = np.where((steps >= 3645) & (steps <= 7956))[0] + 1
goal_btw_1
```

```python
goal_btw_2 = np.where((steps >= 8000) & (steps <= 11778))[0] + 1
goal_btw_2
```

```python
#5th May
#splitting the array to weeks using the "np.array_split()
weeks = np.array_split(steps, 4)
print(weeks)

print('Weekly Total Steps:')
week_total = [print(week.sum()) for week in weeks]

print('\nWeekly Average Steps:')
week_avg = [print(week.mean()) for week in weeks]
```

```python
#Finding outliers(Anomaly) 
mean_step = np.mean(steps)
std_step = np.std(steps)

lower_bound = mean_step - 2*std_step
upper_bound = mean_step + 2*std_step

print(round(lower_bound, 2))
print(round(upper_bound, 2))

outliers = np.where((steps < lower_bound) | (steps > upper_bound))[0] + 1

print('Anomaly Detected:', outliers)

```

The median is 7,946, a bit higher than the mean, which suggests most days are on the more active side.

The lowest day is 3,645 steps, likely a low-movement day, while the highest day reaches 11,778 marking the most active day of the month. The standard deviation is 2,309, showing that the activity varies a lot, some days are very active, others much lighter.
